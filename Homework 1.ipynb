```python
# Create X
X = Asia_cars[['vehicle_weight', 'model_year']].iloc[:7].values
print(X)
print("Shape of X:", X.shape)
```

```python
import numpy as np

# 1. Compute X^T X
XTX = X.T @ X

# 2. Invert XTX
XTX_inv = np.linalg.inv(XTX)

# 3. Create the target array y
y = np.array([1100, 1300, 800, 900, 1000, 1100, 1200])

# 4. Compute weights: w = (X^T X)^(-1) @ X^T @ y
w = XTX_inv @ X.T @ y

# 5. Inspect the weights and their sum
print("Weights (w):", w)
print("Sum of weights:", w.sum())
```
